```python
# Video-level grouping, motion features, and time-series visualizations
# Run this cell after loading `df` above. All geometric measures below use normalized coordinates.
from pathlib import Path
from IPython.display import display

sns.set_theme(style='whitegrid', context='notebook')

# --- 1. Clean, order, and identify the pose columns --------------------------------
df = df.copy().sort_values(['video_name', 'frame_idx']).reset_index(drop=True)
keypoints = [c[:-7] for c in df.columns if c.endswith('_x_norm')]
confidence_cols = [f'{kp}_conf' for kp in keypoints if f'{kp}_conf' in df.columns]

if not keypoints:
    raise ValueError('No normalized keypoint columns were found. Expected columns such as nose_x_norm.')

# A frame-quality score helps identify frames where pose detection may be unreliable.
df['mean_keypoint_confidence'] = df[confidence_cols].mean(axis=1)
df['low_confidence_fraction'] = (df[confidence_cols] < 0.30).mean(axis=1)

# --- 2. Derive frame-level movement and body-shape features -------------------------
by_video = df.groupby('video_name', sort=False)
dt = by_video['timestamp_sec'].diff().where(lambda s: s > 0)
speed_columns = []
for kp in keypoints:
    x_col, y_col = f'{kp}_x_norm', f'{kp}_y_norm'
    dx = by_video[x_col].diff()
    dy = by_video[y_col].diff()
    speed_col = f'{kp}_speed'
    # Units: normalized-image distance per second. The first frame has no previous speed.
    df[speed_col] = np.hypot(dx, dy).div(dt).replace([np.inf, -np.inf], np.nan)
    speed_columns.append(speed_col)

df['mean_joint_speed'] = df[speed_columns].mean(axis=1)
df['wrist_separation'] = np.hypot(
    df['left_wrist_x_norm'] - df['right_wrist_x_norm'],
    df['left_wrist_y_norm'] - df['right_wrist_y_norm'],
)
df['shoulder_width'] = np.hypot(
    df['left_shoulder_x_norm'] - df['right_shoulder_x_norm'],
    df['left_shoulder_y_norm'] - df['right_shoulder_y_norm'],
)

# --- 3. Group frames into videos, then group videos into classes --------------------
video_summary = (
    df.groupby(['video_name', 'class_name'], as_index=False)
      .agg(
          n_frames=('frame_idx', 'size'),
          first_timestamp=('timestamp_sec', 'min'),
          last_timestamp=('timestamp_sec', 'max'),
          mean_confidence=('mean_keypoint_confidence', 'mean'),
          low_confidence_share=('low_confidence_fraction', 'mean'),
          mean_motion=('mean_joint_speed', 'mean'),
          peak_motion=('mean_joint_speed', 'max'),
          mean_wrist_separation=('wrist_separation', 'mean'),
          mean_shoulder_width=('shoulder_width', 'mean'),
      )
)
video_summary['duration_sec'] = video_summary['last_timestamp'] - video_summary['first_timestamp']
video_summary['fps_estimate'] = (
    (video_summary['n_frames'] - 1) / video_summary['duration_sec'].replace(0, np.nan)
)

# Motion groups make it easy to select quiet, typical, or expressive videos.
motion_rank = video_summary['mean_motion'].rank(method='first')
n_motion_groups = min(3, len(video_summary))
video_summary['motion_group'] = (
    pd.qcut(motion_rank, q=n_motion_groups, labels=['low', 'medium', 'high'][-n_motion_groups:])
    if n_motion_groups > 1 else 'only video'
)

class_summary = (
    video_summary.groupby('class_name', as_index=False)
    .agg(
        n_videos=('video_name', 'size'),
        total_frames=('n_frames', 'sum'),
        median_duration_sec=('duration_sec', 'median'),
        mean_confidence=('mean_confidence', 'mean'),
        mean_motion=('mean_motion', 'mean'),
    )
    .sort_values('n_videos', ascending=False)
)

# Dictionary form: use video_groups['a_class'] to retrieve all of that class's video summaries.
video_groups = {name: group.reset_index(drop=True)
                for name, group in video_summary.groupby('class_name', sort=True)}

print(f'{len(df):,} frames | {len(video_summary):,} videos | {len(class_summary):,} classes')
display(video_summary.sort_values(['class_name', 'video_name']).head(10))
display(class_summary.head(10))

# --- 4. Dataset and video-group visualizations --------------------------------------
fig, axes = plt.subplots(1, 3, figsize=(22, 6))
sns.barplot(data=class_summary, x='class_name', y='n_videos', color='#4C78A8', ax=axes[0])
axes[0].set(title='Videos per class', xlabel='Class', ylabel='Number of videos')
axes[0].tick_params(axis='x', rotation=90, labelsize=8)

sns.boxplot(data=video_summary, x='motion_group', y='mean_motion', hue='motion_group',
            palette='viridis', legend=False, ax=axes[1])
axes[1].set(title='Video groups by movement', xlabel='Motion group',
            ylabel='Mean normalized joint speed')

sns.scatterplot(data=video_summary, x='duration_sec', y='mean_motion', hue='mean_confidence',
                size='n_frames', sizes=(20, 220), palette='viridis', alpha=.8, ax=axes[2])
axes[2].set(title='Each point is one video', xlabel='Duration (seconds)',
            ylabel='Mean normalized joint speed')
plt.tight_layout()
plt.show()

# --- 5. Inspect one video's time series ---------------------------------------------
def plot_video_time_series(video_name, data=None, save_path=None):
    """Plot wrist coordinates, joint speed, and confidence for one video.

    Parameters
    ----------
    video_name : str
        Exact video name, video basename, or a unique case-insensitive substring.
    data : pandas.DataFrame, optional
        Coordinate data. Defaults to the notebook's `df` DataFrame.
    save_path : str or pathlib.Path, optional
        Save the figure to this path instead of only displaying it.

    Returns
    -------
    tuple[matplotlib.figure.Figure, pandas.DataFrame]
        The plotted figure and the selected video's frame data.
    """
    if data is None:
        data = df

    if 'video_name' not in data.columns:
        raise KeyError("The data must contain a 'video_name' column.")

    available = data['video_name'].dropna().astype(str).drop_duplicates()
    name = str(video_name).strip()
    exact = available[available == name]

    if exact.empty:
        folded = available.str.lower()
        matches = available[folded == name.lower()].tolist()
        if not matches:
            matches = available[
                folded.str.contains(name.lower(), regex=False)
                | available.apply(lambda value: Path(value).stem.lower() == name.lower())
            ].tolist()

        if len(matches) == 0:
            raise ValueError(f"Video '{video_name}' was not found in the data.")
        if len(matches) > 1:
            preview = '\n'.join(matches[:10])
            raise ValueError(
                f"Video name '{video_name}' is ambiguous ({len(matches)} matches). "
                f"Use an exact name, such as:\n{preview}"
            )
        resolved_name = matches[0]
    else:
        resolved_name = exact.iloc[0]

    one_video = data.loc[data['video_name'].astype(str).eq(resolved_name)].copy()
    one_video = one_video.sort_values(
        'timestamp_sec' if 'timestamp_sec' in one_video else 'frame_idx'
    )

    if 'mean_joint_speed' not in one_video:
        keypoints = [c[:-7] for c in one_video.columns if c.endswith('_x_norm')]
        confidence_cols = [f'{kp}_conf' for kp in keypoints if f'{kp}_conf' in one_video]
        one_video['mean_keypoint_confidence'] = one_video[confidence_cols].mean(axis=1)

        dt = one_video['timestamp_sec'].diff().where(lambda s: s > 0)
        speed_columns = []
        for kp in keypoints:
            dx = one_video[f'{kp}_x_norm'].diff()
            dy = one_video[f'{kp}_y_norm'].diff()
            speed_col = f'{kp}_speed'
            one_video[speed_col] = np.hypot(dx, dy).div(dt).replace([np.inf, -np.inf], np.nan)
            speed_columns.append(speed_col)
        one_video['mean_joint_speed'] = one_video[speed_columns].mean(axis=1)

    fig, axes = plt.subplots(3, 1, figsize=(15, 11), sharex=True)
    for wrist, color in [('left_wrist', '#E45756'), ('right_wrist', '#4C78A8')]:
        axes[0].plot(one_video['timestamp_sec'], one_video[f'{wrist}_x_norm'],
                     label=f'{wrist} x', color=color)
        axes[1].plot(one_video['timestamp_sec'], one_video[f'{wrist}_y_norm'],
                     label=f'{wrist} y', color=color)
    axes[2].plot(one_video['timestamp_sec'], one_video['mean_joint_speed'],
                 label='mean joint speed', color='#54A24B')
    axes[2].plot(one_video['timestamp_sec'], one_video['mean_keypoint_confidence'],
                 label='mean confidence', color='#B279A2', alpha=.8)
    axes[0].set(title=f'Time series: {resolved_name}', ylabel='Normalized x')
    axes[1].set(ylabel='Normalized y')
    axes[2].set(xlabel='Time (seconds)', ylabel='Speed / confidence')
    for ax in axes:
        ax.legend(ncol=2)
    plt.tight_layout()

    if save_path:
        fig.savefig(save_path, dpi=150, bbox_inches='tight')

    plt.show()
    return fig, one_video

# Example: pass any exact video name or a unique substring.
VIDEO_TO_PLOT = video_summary.sort_values('n_frames', ascending=False).iloc[0]['video_name']
figure, selected_video = plot_video_time_series(VIDEO_TO_PLOT)

# --- 6. Compare classes on a shared 0-100% timeline --------------------------------
# This removes duration differences before averaging trajectories across videos.
df['relative_video_time'] = by_video['timestamp_sec'].transform(
    lambda t: (t - t.min()) / (t.max() - t.min()) if t.max() > t.min() else 0
)
df['time_bin'] = pd.cut(df['relative_video_time'], bins=np.linspace(0, 1, 21),
                        include_lowest=True, labels=False)
class_trajectory = (
    df.groupby(['class_name', 'time_bin'], as_index=False)
      .agg(mean_motion=('mean_joint_speed', 'mean'),
           mean_left_wrist_y=('left_wrist_y_norm', 'mean'),
           n_frames=('frame_idx', 'size'))
)
class_trajectory['relative_time_pct'] = (class_trajectory['time_bin'] + 0.5) * 5

# Plot the most represented classes so a large label set stays readable.
classes_to_compare = class_summary.head(8)['class_name']
plt.figure(figsize=(14, 6))
sns.lineplot(data=class_trajectory[class_trajectory['class_name'].isin(classes_to_compare)],
             x='relative_time_pct', y='mean_motion', hue='class_name', marker='o')
plt.title('Class motion profiles on a normalized video timeline')
plt.xlabel('Video progress (%)')
plt.ylabel('Mean normalized joint speed')
plt.legend(title='Class', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()

# Optional: save reusable tables after reviewing them.
# Path('Data Product').mkdir(exist_ok=True)
# video_summary.to_csv('Data Product/video_summary.csv', index=False)
# class_summary.to_csv('Data Product/class_summary.csv', index=False)
# class_trajectory.to_csv('Data Product/class_motion_time_series.csv', index=False)
```
