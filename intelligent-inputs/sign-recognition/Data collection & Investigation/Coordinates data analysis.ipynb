```python
# ==============================================================================
# Video-level grouping, motion features, and time-series visualizations
# ==============================================================================

from pathlib import Path
from IPython.display import display

sns.set_theme(style='whitegrid', context='notebook')

# ==============================================================================
# 1. Clean, order, and identify the pose columns
# ==============================================================================

df = df.copy().sort_values(
    ['video_name', 'frame_idx']
).reset_index(drop=True)

keypoints = [
    c[:-7] for c in df.columns if c.endswith('_x_norm')
]

confidence_cols = [
    f'{kp}_conf' for kp in keypoints
    if f'{kp}_conf' in df.columns
]

if not keypoints:
    raise ValueError(
        'No normalized keypoint columns were found. '
        'Expected columns such as nose_x_norm.'
    )

# Frame-quality scores
df['mean_keypoint_confidence'] = df[confidence_cols].mean(axis=1)

df['low_confidence_fraction'] = (
    df[confidence_cols] < 0.30
).mean(axis=1)


# ==============================================================================
# 2. Derive frame-level movement and body-shape features
# ==============================================================================

by_video = df.groupby('video_name', sort=False)

dt = by_video['timestamp_sec'].diff().where(
    lambda s: s > 0
)

speed_columns = []

for kp in keypoints:

    x_col = f'{kp}_x_norm'
    y_col = f'{kp}_y_norm'

    dx = by_video[x_col].diff()
    dy = by_video[y_col].diff()

    speed_col = f'{kp}_speed'

    df[speed_col] = (
        np.hypot(dx, dy)
        .div(dt)
        .replace([np.inf, -np.inf], np.nan)
    )

    speed_columns.append(speed_col)

df['mean_joint_speed'] = df[speed_columns].mean(axis=1)

df['wrist_separation'] = np.hypot(
    df['left_wrist_x_norm'] - df['right_wrist_x_norm'],
    df['left_wrist_y_norm'] - df['right_wrist_y_norm']
)

df['shoulder_width'] = np.hypot(
    df['left_shoulder_x_norm'] - df['right_shoulder_x_norm'],
    df['left_shoulder_y_norm'] - df['right_shoulder_y_norm']
)


# ==============================================================================
# 3. Group frames into videos, then group videos into classes
# ==============================================================================

video_summary = (
    df.groupby(['video_name', 'class_name'], as_index=False)
    .agg(
        n_frames=('frame_idx', 'size'),
        first_timestamp=('timestamp_sec', 'min'),
        last_timestamp=('timestamp_sec', 'max'),
        mean_confidence=('mean_keypoint_confidence', 'mean'),
        low_confidence_share=('low_confidence_fraction', 'mean'),
        mean_motion=('mean_joint_speed', 'mean'),
        peak_motion=('mean_joint_speed', 'max'),
        mean_wrist_separation=('wrist_separation', 'mean'),
        mean_shoulder_width=('shoulder_width', 'mean')
    )
)

video_summary['duration_sec'] = (
    video_summary['last_timestamp']
    - video_summary['first_timestamp']
)

video_summary['fps_estimate'] = (
    (video_summary['n_frames'] - 1)
    / video_summary['duration_sec'].replace(0, np.nan)
)

# Motion groups
motion_rank = video_summary['mean_motion'].rank(method='first')

n_motion_groups = min(3, len(video_summary))

video_summary['motion_group'] = (
    pd.qcut(
        motion_rank,
        q=n_motion_groups,
        labels=['low', 'medium', 'high'][-n_motion_groups:]
    )
    if n_motion_groups > 1 else 'only video'
)

# Class-level summary
class_summary = (
    video_summary.groupby('class_name', as_index=False)
    .agg(
        n_videos=('video_name', 'size'),
        total_frames=('n_frames', 'sum'),
        median_duration_sec=('duration_sec', 'median'),
        mean_confidence=('mean_confidence', 'mean'),
        mean_motion=('mean_motion', 'mean')
    )
    .sort_values('n_videos', ascending=False)
)

# Dictionary containing video summaries for each class
video_groups = {
    name: group.reset_index(drop=True)
    for name, group in video_summary.groupby(
        'class_name', sort=True
    )
}

print(
    f'{len(df):,} frames | '
    f'{len(video_summary):,} videos | '
    f'{len(class_summary):,} classes'
)

display(
    video_summary.sort_values(
        ['class_name', 'video_name']
    ).head(10)
)

display(class_summary.head(10))


# ==============================================================================
# 4. Dataset and video-group visualizations
# ==============================================================================

fig, axes = plt.subplots(1, 3, figsize=(22, 6))

sns.barplot(
    data=class_summary,
    x='class_name',
    y='n_videos',
    color='#4C78A8',
    ax=axes[0]
)

axes[0].set(
    title='Videos per class',
    xlabel='Class',
    ylabel='Number of videos'
)

axes[0].tick_params(
    axis='x',
    rotation=90,
    labelsize=8
)

sns.boxplot(
    data=video_summary,
    x='motion_group',
    y='mean_motion',
    hue='motion_group',
    palette='viridis',
    legend=False,
    ax=axes[1]
)

axes[1].set(
    title='Video groups by movement',
    xlabel='Motion group',
    ylabel='Mean normalized joint speed'
)

sns.scatterplot(
    data=video_summary,
    x='duration_sec',
    y='mean_motion',
    hue='mean_confidence',
    size='n_frames',
    sizes=(20, 220),
    palette='viridis',
    alpha=.8,
    ax=axes[2]
)

axes[2].set(
    title='Each point is one video',
    xlabel='Duration (seconds)',
    ylabel='Mean normalized joint speed'
)

plt.tight_layout()
plt.show()


# ==============================================================================
# 5. Function to plot an individual video's time series
# ==============================================================================

def plot_video_time_series(
    video_name,
    data=None,
    save_path=None,
    show_plot=True
):
    """
    Plot all keypoint coordinates, joint speed, and confidence for one video.

    Parameters
    ----------
    video_name : str
        Exact video name, basename, or unique case-insensitive substring.

    data : pandas.DataFrame, optional
        Coordinate data. Defaults to the notebook's df DataFrame.

    save_path : str or pathlib.Path, optional
        Save the figure to this path.

    show_plot : bool, default=True
        True  -> Display the figure in the notebook.
        False -> Do not display the figure.

    Returns
    -------
    tuple
        The plotted figure and the selected video's frame data.
    """

    if data is None:
        data = df

    if 'video_name' not in data.columns:
        raise KeyError(
            "The data must contain a 'video_name' column."
        )

    available = (
        data['video_name']
        .dropna()
        .astype(str)
        .drop_duplicates()
    )

    name = str(video_name).strip()

    exact = available[available == name]

    # Resolve video name
    if exact.empty:

        folded = available.str.lower()

        matches = available[
            folded == name.lower()
        ].tolist()

        if not matches:
            matches = available[
                folded.str.contains(name.lower(), regex=False)
                | available.apply(
                    lambda value:
                    Path(value).stem.lower() == name.lower()
                )
            ].tolist()

        if len(matches) == 0:
            raise ValueError(
                f"Video '{video_name}' was not found in the data."
            )

        if len(matches) > 1:

            preview = '\n'.join(matches[:10])

            raise ValueError(
                f"Video name '{video_name}' is ambiguous "
                f"({len(matches)} matches). "
                f"Use an exact name:\n{preview}"
            )

        resolved_name = matches[0]

    else:
        resolved_name = exact.iloc[0]

    # Select video frames
    one_video = data.loc[
        data['video_name'].astype(str).eq(resolved_name)
    ].copy()

    one_video = one_video.sort_values(
        'timestamp_sec'
        if 'timestamp_sec' in one_video.columns
        else 'frame_idx'
    )

    # Identify all detected keypoints
    keypoints = [
        c[:-7]
        for c in one_video.columns
        if c.endswith('_x_norm')
    ]

    if not keypoints:
        raise ValueError(
            'No normalized keypoint columns were found. '
            'Expected columns such as nose_x_norm.'
        )

    # Calculate missing motion features if necessary
    if 'mean_joint_speed' not in one_video.columns:

        confidence_cols = [
            f'{kp}_conf'
            for kp in keypoints
            if f'{kp}_conf' in one_video.columns
        ]

        one_video['mean_keypoint_confidence'] = (
            one_video[confidence_cols].mean(axis=1)
        )

        dt = one_video['timestamp_sec'].diff().where(
            lambda s: s > 0
        )

        speed_columns = []

        for kp in keypoints:

            dx = one_video[f'{kp}_x_norm'].diff()
            dy = one_video[f'{kp}_y_norm'].diff()

            speed_col = f'{kp}_speed'

            one_video[speed_col] = (
                np.hypot(dx, dy)
                .div(dt)
                .replace([np.inf, -np.inf], np.nan)
            )

            speed_columns.append(speed_col)

        one_video['mean_joint_speed'] = (
            one_video[speed_columns].mean(axis=1)
        )

    # Create figure
    fig, axes = plt.subplots(
        3, 1,
        figsize=(18, 11),
        sharex=True
    )

    # Plot every detected keypoint's normalized X and Y coordinates
    keypoint_colors = sns.color_palette(
        'husl',
        n_colors=len(keypoints)
    )

    for kp, color in zip(keypoints, keypoint_colors):

        axes[0].plot(
            one_video['timestamp_sec'],
            one_video[f'{kp}_x_norm'],
            label=kp,
            color=color,
            alpha=.85
        )

        axes[1].plot(
            one_video['timestamp_sec'],
            one_video[f'{kp}_y_norm'],
            label=kp,
            color=color,
            alpha=.85
        )

    # Joint speed
    axes[2].plot(
        one_video['timestamp_sec'],
        one_video['mean_joint_speed'],
        label='mean joint speed',
        color='#54A24B'
    )

    # Keypoint confidence
    axes[2].plot(
        one_video['timestamp_sec'],
        one_video['mean_keypoint_confidence'],
        label='mean confidence',
        color='#B279A2',
        alpha=.8
    )

    # Labels and titles
    fig.suptitle(f'Pose time series: {resolved_name}')

    axes[0].set(
        title='All keypoints: normalized X coordinates',
        ylabel='Normalized x'
    )

    axes[1].set(
        title='All keypoints: normalized Y coordinates',
        ylabel='Normalized y'
    )

    axes[2].set(
        xlabel='Time (seconds)',
        ylabel='Speed / confidence'
    )

    fig.legend(
        handles=axes[0].get_lines(),
        labels=keypoints,
        title='Keypoint',
        loc='center left',
        bbox_to_anchor=(.89, .5)
    )

    axes[2].legend(ncol=2)

    plt.tight_layout(rect=(0, 0, .87, .96))

    # Save figure
    if save_path is not None:

        save_path = Path(save_path)

        save_path.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        fig.savefig(
            save_path,
            dpi=150,
            bbox_inches='tight'
        )

    # Display or close figure
    if show_plot:
        plt.show()
    else:
        plt.close(fig)

    return fig, one_video


# ==============================================================================
# 6. Inspect one video's time series
# ==============================================================================

VIDEO_TO_PLOT = (
    video_summary
    .sort_values('n_frames', ascending=False)
    .iloc[0]['video_name']
)

figure, selected_video = plot_video_time_series(
    video_name=VIDEO_TO_PLOT,
    data=df,
    show_plot=True
)


# ==============================================================================
# 7. Compare classes on a shared 0-100% timeline
# ==============================================================================

df['relative_video_time'] = (
    by_video['timestamp_sec'].transform(
        lambda t: (
            (t - t.min()) / (t.max() - t.min())
            if t.max() > t.min() else 0
        )
    )
)

df['time_bin'] = pd.cut(
    df['relative_video_time'],
    bins=np.linspace(0, 1, 21),
    include_lowest=True,
    labels=False
)

class_trajectory = (
    df.groupby(
        ['class_name', 'time_bin'],
        as_index=False
    )
    .agg(
        mean_motion=('mean_joint_speed', 'mean'),
        mean_left_wrist_y=('left_wrist_y_norm', 'mean'),
        n_frames=('frame_idx', 'size')
    )
)

class_trajectory['relative_time_pct'] = (
    (class_trajectory['time_bin'] + 0.5) * 5
)

# Compare the most represented classes
classes_to_compare = class_summary.head(8)['class_name']

plt.figure(figsize=(14, 6))

sns.lineplot(
    data=class_trajectory[
        class_trajectory['class_name'].isin(classes_to_compare)
    ],
    x='relative_time_pct',
    y='mean_motion',
    hue='class_name',
    marker='o'
)

plt.title(
    'Class motion profiles on a normalized video timeline'
)

plt.xlabel('Video progress (%)')
plt.ylabel('Mean normalized joint speed')

plt.legend(
    title='Class',
    bbox_to_anchor=(1.02, 1),
    loc='upper left'
)

plt.tight_layout()
plt.show()


# ==============================================================================
# 8. Save individual video plots organized by class
# ==============================================================================

SAVE_INDIVIDUAL_PLOTS = True

OUTPUT_DIR = Path('plots')

if SAVE_INDIVIDUAL_PLOTS:

    OUTPUT_DIR.mkdir(
        parents=True,
        exist_ok=True
    )

    total_saved = 0

    for class_name, group in video_summary.groupby('class_name'):

        # Create one folder per class
        class_dir = OUTPUT_DIR / str(class_name)

        class_dir.mkdir(
            parents=True,
            exist_ok=True
        )

        for video_name in group['video_name']:

            plot_path = (
                class_dir / f'{Path(str(video_name)).stem}.png'
            )

            plot_video_time_series(
                video_name=video_name,
                data=df,
                save_path=plot_path,
                show_plot=False
            )

            total_saved += 1

        print(
            f"Class '{class_name}': "
            f"{len(group)} diagrams saved."
        )

    print("=" * 50)
    print(f"Total diagrams saved: {total_saved}")
    print(f"Output directory: {OUTPUT_DIR.resolve()}")
    print("=" * 50)


# ==============================================================================
# 9. Optional: Save reusable tables
# ==============================================================================

# OUTPUT_DIR.mkdir(exist_ok=True)

# video_summary.to_csv(
#     OUTPUT_DIR / 'video_summary.csv',
#     index=False
# )

# class_summary.to_csv(
#     OUTPUT_DIR / 'class_summary.csv',
#     index=False
# )

# class_trajectory.to_csv(
#     OUTPUT_DIR / 'class_motion_time_series.csv',
#     index=False
# )
```
